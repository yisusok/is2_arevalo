## Ejercicio 1: Ingeniería Financiera de un Proyecto de Software

### Premisa:
- Inversión mensual: $1000/mes durante 12 meses (pagado mensual vencido).
- Flujos subyacentes al mes 12 con Valor Presente ($PV_{12}$) de $18,000.
- Tasa efectiva mensual $r = 1\% = 0.01$.


```python
r = 0.01
costo_mensual = 1000.0

# a. Plan original (12 meses)
pv_costos_a = sum(costo_mensual / ((1 + r)**t) for t in range(1, 13))
pv_beneficio_a = 18000.0 / ((1 + r)**12)
npv_a = pv_beneficio_a - pv_costos_a
rent_a = npv_a / pv_costos_a

# b. Retraso de 3 meses (15 meses)
pv_costos_b = sum(costo_mensual / ((1 + r)**t) for t in range(1, 16))
pv_beneficio_b = 18000.0 / ((1 + r)**15)
npv_b = pv_beneficio_b - pv_costos_b
rent_b = npv_b / pv_costos_b

# d. Retraso de 6 meses (18 meses)
pv_costos_d = sum(costo_mensual / ((1 + r)**t) for t in range(1, 19))
pv_beneficio_d = 18000.0 / ((1 + r)**18)
npv_d = pv_beneficio_d - pv_costos_d
rent_d = npv_d / pv_costos_d

# e. PM Profesional (+5% costo = $1050/mes, 12 meses)
pv_costos_e = sum(1050.0 / ((1 + r)**t) for t in range(1, 13))
npv_e = pv_beneficio_a - pv_costos_e
rent_e = npv_e / pv_costos_e

# f. Desafío 15 meses conservando NPV original ($4719.01)
pv_costos_f_permitido = pv_beneficio_b - npv_a
factor_15 = sum(1.0 / ((1 + r)**t) for t in range(1, 16))
c_max = pv_costos_f_permitido / factor_15

print(f'a. NPV Plan (12m): ${npv_a:.2f} | Rentabilidad: {rent_a*100:.2f}%')
print(f'b. NPV Retraso 3m (15m): ${npv_b:.2f} | Rentabilidad: {rent_b*100:.2f}%')
print(f'd. NPV Retraso 6m (18m): ${npv_d:.2f} | Rentabilidad: {rent_d*100:.2f}%')
print(f'e. NPV PM Certificado (12m, +5%): ${npv_e:.2f} | Rentabilidad: {rent_e*100:.2f}%')
print(f'f. Costo Mensual Máximo (15m, NPV objetivo): ${c_max:.2f}/mes (Reducción del {((1000-c_max)/1000)*100:.2f}%)')
```

```python
# Gráfico de Evolución del NPV según meses de retraso
meses = list(range(12, 25))
npvs = []
for m in meses:
    c_pv = sum(1000.0 / ((1 + r)**t) for t in range(1, m + 1))
    b_pv = 18000.0 / ((1 + r)**m)
    npvs.append(b_pv - c_pv)

plt.figure(figsize=(10, 5))
plt.plot(meses, npvs, marker='o', color='#1f77b4', linewidth=2, label='NPV del Proyecto ($)')
plt.axhline(0, color='red', linestyle='--', linewidth=1, label='Punto de Inviabilidad (NPV = 0)')
plt.axvline(12, color='green', linestyle=':', label='Plan Original (12 meses)')
plt.scatter([15, 18], [npv_b, npv_d], color=['orange', 'red'], s=80, zorder=5)
plt.title('Destrucción de Valor Financiero por Retraso en la Entrega (r = 1% TEM)', fontsize=13, fontweight='bold')
plt.xlabel('Plazo Total de Ejecución (Meses)')
plt.ylabel('Valor Presente Neto - NPV ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

## Ejercicio 2: Equilibrio Financiero y Escenarios de Negociación

### Premisa:
- Duración: 12 meses, staff $1000/mes.
- Costo de oportunidad: $r = 1\%$ TEM.
- Pago contractual a la entrega (mes 12): $14,000.


```python
pv_costos_base = sum(1000.0 / ((1 + r)**t) for t in range(1, 13))
pv_ingreso_base = 14000.0 / ((1 + r)**12)
npv_proveedor_base = pv_ingreso_base - pv_costos_base

# Escenario a: Pago en mes 14
pago_m14 = pv_ingreso_base * ((1 + r)**14)

# Escenario b: Entrega mes 6, pago mes 12
pv_costos_6 = sum(1000.0 / ((1 + r)**t) for t in range(1, 7))
pv_ingreso_req_b = pv_costos_6 + npv_proveedor_base
pago_m12_b = pv_ingreso_req_b * ((1 + r)**12)

# Escenario c: Entrega mes 6, pago mes 6
pago_m6_c = pv_ingreso_req_b * ((1 + r)**6)

print(f'Línea de Base NPV Proveedor: ${npv_proveedor_base:.2f}')
print(f'Escenario a. Pago requerido en Mes 14: ${pago_m14:.2f}')
print(f'Escenario b. Pago requerido en Mes 12 (Entrega Mes 6): ${pago_m12_b:.2f}')
print(f'Escenario c. Pago requerido en Mes 6 (Entrega Mes 6): ${pago_m6_c:.2f}')
```

## Ejercicio 14: Diagrama de Red, PERT/CPM y Nivelación de Recursos (Staff = 2)

### Red de Tareas:
- **A**: Requerimientos (dur=5)
- **B**: Arquitectura (dur=3, dep=A)
- **C**: Diseño (dur=10, dep=A)
- **D**: Test cases (dur=20, dep=A)
- **E**: Programa 1 (dur=5, dep=B,C)
- **F**: Programa 2 (dur=6, dep=B,C)
- **G**: Programa 3 (dur=7, dep=B,C)
- **H**: Test F1 (dur=10, dep=E,F,D)
- **I**: Test F2 (dur=9, dep=G,D)
- **J**: System Test (dur=12, dep=H,I)


```python
tasks = {
    'A': {'dur': 5, 'deps': []},
    'B': {'dur': 3, 'deps': ['A']},
    'C': {'dur': 10, 'deps': ['A']},
    'D': {'dur': 20, 'deps': ['A']},
    'E': {'dur': 5, 'deps': ['B', 'C']},
    'F': {'dur': 6, 'deps': ['B', 'C']},
    'G': {'dur': 7, 'deps': ['B', 'C']},
    'H': {'dur': 10, 'deps': ['E', 'F', 'D']},
    'I': {'dur': 9, 'deps': ['G', 'D']},
    'J': {'dur': 12, 'deps': ['H', 'I']}
}

# CPM Infinito
ES, EF = {}, {}
for t_id, d in tasks.items():
    ES[t_id] = 0 if not d['deps'] else max(EF[dep] for dep in d['deps'])
    EF[t_id] = ES[t_id] + d['dur']
dur_inf = max(EF.values())

LF, LS = {}, {}
for t_id in reversed(list(tasks.keys())):
    succs = [s for s, data in tasks.items() if t_id in data['deps']]
    LF[t_id] = dur_inf if not succs else min(LS[s] for s in succs)
    LS[t_id] = LF[t_id] - tasks[t_id]['dur']

float_t = {t_id: LS[t_id] - ES[t_id] for t_id in tasks}
crit_inf = [t_id for t_id in tasks if float_t[t_id] == 0]

print(f'Parte a: Duración Recursos Infinitos = {dur_inf} días')
print(f'Camino Crítico: {" -> ".join(crit_inf)}')

# Nivelación 2 Personas
time, completed, in_progress = 0, set(), {}
st_2p, et_2p = {}, {}
all_t = set(tasks.keys())
while len(completed) < len(tasks):
    for t in [k for k, v in in_progress.items() if v == time]:
        completed.add(t)
        del in_progress[t]
    ready = [t for t in all_t if t not in completed and t not in in_progress and all(dep in completed for dep in tasks[t]['deps'])]
    ready.sort(key=lambda x: (LS[x], float_t[x], x))
    for t_s in ready[:2 - len(in_progress)]:
        st_2p[t_s] = time
        et_2p[t_s] = time + tasks[t_s]['dur']
        in_progress[t_s] = et_2p[t_s]
    if in_progress:
        time = min(in_progress.values())
    else:
        break

dur_2p = max(et_2p.values())
print(f'Parte b: Duración 2 Personas = {dur_2p} días')
```

```python
# Gráfico Gantt Comparativo (Recursos Infinitos vs 2 Personas)
fig, ax = plt.subplots(figsize=(12, 6))

y_pos = np.arange(len(tasks))
task_list = list(tasks.keys())

# Recursos Infinitos
for i, t_id in enumerate(task_list):
    ax.barh(i - 0.2, tasks[t_id]['dur'], left=ES[t_id], height=0.35, color='#2ca02c' if float_t[t_id]==0 else '#aec7e8', align='center', label='Infinito (Crítica)' if float_t[t_id]==0 and i==0 else ('Infinito (No crítica)' if i==0 else ''))

# 2 Personas
for i, t_id in enumerate(task_list):
    ax.barh(i + 0.2, tasks[t_id]['dur'], left=st_2p[t_id], height=0.35, color='#d62728', align='center', alpha=0.85, label='Staff = 2 personas' if i==0 else '')

ax.set_yticks(y_pos)
ax.set_yticklabels([f"{t} ({tasks[t]['dur']}d)" for t in task_list])
ax.invert_yaxis()
ax.set_xlabel('Días Calendario')
ax.set_title('Comparación de Cronogramas Gantt: Recursos Infinitos (47d) vs 2 Personas (52d)', fontsize=12, fontweight='bold')
ax.axvline(47, color='green', linestyle='--', label='Fin Rec. Infinitos (47d)')
ax.axvline(52, color='red', linestyle='--', label='Fin 2 Personas (52d)')
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
